```python
x=digits.data
y=digits.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.2)
```

```python
model.fit(X_train,y_train) 
```

```python
out=model.predict(X_test)
```

```python
sahih=0
qalat=0

for a in er:
    if(a==0):
        sahih=sahih+1
    if (a!=0):
        qalat=qalat+1

print(sahih)
print(qalat)
```
